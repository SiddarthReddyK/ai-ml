```python
scores = {
    "accuracy": 0.91,
    "precision": 0.87,
    "recall": 0.84
}
print(scores["accuracy"]) #Dictionaries are particularly useful when lookup by a meaningful identifier is more important than maintaining positional order.
```

```python
regions = [
    "South", "North", "South", "West",
    "East", "South", "North", "West", "South"
] #list

region_counts = {}

for region in regions:
  region_counts[region] = region_counts.get(region, 0)+1

print(region_counts)
```

```python
students = [
    {"id": 101, "name": "Aarav", "cgpa": 8.9},
    {"id": 102, "name": "Meera", "cgpa": 9.2},
    {"id": 103, "name": "Rohan", "cgpa": 8.4}
]

students_by_id = {}

for student in students:
    students_by_id[student["id"]] = student

print(students_by_id[102])
```

```python
results = [
    ("Logistic Regression", 0.96),
    ("Random Forest", 0.91),
    ("SVM", 0.88),
    ("XGBoost", 0.94)
]

model_scores = dict(results)

best_model = max(model_scores, key=model_scores.get)

print(best_model)
print(model_scores[best_model])
```
