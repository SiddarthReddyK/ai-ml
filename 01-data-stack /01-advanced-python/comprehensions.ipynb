**List Comprehensions**

A list comprehension is a concise way of creating a new list from an existing iterable, often with filtering or transformation.

1.Feature preprocessing

```python
temperatures = [18.5, 21.2, 19.8, 25.4, 30.1]

fahrenheit = [
    (temp*9/5) + 32
    for temp in temperatures
]
print(fahrenheit)
```

2.Filtering data

```python
confidence_scores = [
    0.42, 0.91, 0.78, 0.35, 0.96, 0.83
]

high_confidence = [
    score
    for score in confidence_scores
    if score >= 0.8
]
print(high_confidence)
```

```python
#general structure

[expression for item in iterable if condition]
```

**Dictionary Comprehension**

Dictionary comprehensions work similarly, but produce: key->value


1.Creating a feature mapping

This kind of mapping is useful when working with feature vectors and preprocessing pipelines.

```python
features = [
    "age",
    "income",
    "credit_score",
    "account_balance"
]

feature_index = {
    feature : index
    for index, feature in enumerate(features)
}
print(feature_index)
```

2.Filtering model results


```python
model_scores = {
    "Logistic Regression": 0.84,
    "Random Forest": 0.91,
    "SVM": 0.88,
    "XGBoost": 0.94,
    "Naive Bayes": 0.79
}
strong_models = {
    model: score
    for model, score in model_scores.items()
    if score >= 0.85
}

print(strong_models)
```

3.Transforming dictionary values

```python
features = {
    "age" : 19,
    "experience" : 1,
    "projects" : 4
}

normalized = {
    feature: value/10
    for feature, value in features.items()
}
print(normalized)
```

4.Nested Comprehensions

```python
matrix = [
    [1,2,3],
    [4,5,6],
    [7,8,9]
]

flattened = [
    value
    for row in matrix
    for value in row
]
print(flattened)
```

```python
model_scores.keys() #for keys
model_scores.values() #for values
model_scores.items() #for both
#also
for model, score in model_scores.items():
  print(model,score)
```

```python
#to check key existence
if "XGBoost" in model_scores:
    print(model_scores["XGBoost"])
```

```python
skills = [
    "Python", "SQL", "Python", "ML",
    "SQL", "Python", "Deep Learning",
    "ML", "Python"
]
skill_counts = {
    skill : skills.count(skill)
    for skill in set(skills)
}
print(skill_counts)
```
